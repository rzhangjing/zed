## A session

A `Session` is a project open in memory. Every action is a command with an id and parameters, the same ids the app's ribbon, the command line and the AI Assistant use.

```python
import rustgis

s = rustgis.Session(sample=True)  # the World Atlas sample
[layer["name"] for layer in s.layers()]
```

```python
s.run("gp.run", tool="buffer", params={"input": "Cities", "distance": 100})
s.run("layer.symbology", layer="Countries", type="graduated", field="pop_est")
s  # a session on the last line shows its map
```

Layers come out as GeoJSON or GeoDataFrames, and go back in the same way.

```python
gdf = s.layer("Cities").to_geopandas()
big = gdf[gdf.pop_max > 10_000_000]
big[["name", "pop_max"]].sort_values("pop_max", ascending=False).head()
```

## The app in the notebook

`rustgis.Map` shows the RustGIS app (ribbon, panes and map) below. What Python adds or styles appears in the app, and what you do in the app (a pan, an edit, a new layer) comes back to Python.

```python
m = rustgis.Map(sample=True)
m
```

```python
m.add_geopandas(big, name="Megacities")
m.run("layer.symbology", layer="Countries", type="graduated", field="pop_est")
m.set_center(110, 25, scale=30_000_000)
```

Pan or zoom the map above, or edit a layer, then read the state back:

```python
m.info()["center"], m.info()["scale"], [layer["name"] for layer in m.layers()]
```

```python
m.run_in_app("ui.language", code="de")  # commands only the app has (ui.*)
```

## Save

The project opens in the RustGIS desktop app and the [web app](https://web.rustgis.app) as it is.

```python
m.save("atlas.rgis")
m.to_png("atlas.png", width=1600, height=900);
```
